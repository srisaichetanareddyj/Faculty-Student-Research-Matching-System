```python
# Combine student research interests and faculty publications

publication_texts = (
    list(students_df["research_interests"]) +
    list(faculty_df["publications"])
)

# Create TF-IDF vectors
publication_vectorizer = TfidfVectorizer()

publication_tfidf = publication_vectorizer.fit_transform(
    publication_texts
)

# Separate student and faculty vectors
num_students = len(students_df)

student_publication_vectors = publication_tfidf[:num_students]
faculty_publication_vectors = publication_tfidf[num_students:]

# Calculate publication similarity
publication_similarity = cosine_similarity(
    student_publication_vectors,
    faculty_publication_vectors
)

print(publication_similarity)
```

```python
# Store final recommendations
final_recommendations = []

# Go through each student
for student in final_similarity_df.index:

    # Get the top 3 faculty scores for this student
    top_3 = final_similarity_df.loc[student].sort_values(
        ascending=False
    ).head(3)

    # Go through the top 3 matches
    for rank, (faculty, score) in enumerate(top_3.items(), start=1):

        # Save each recommendation
        final_recommendations.append({
            "Student": student,
            "Rank": rank,
            "Recommended Faculty": faculty,
            "Final Match Score": round(score, 3)
        })

# Convert the results into a DataFrame
final_recommendations_df = pd.DataFrame(final_recommendations)

# Display the recommendations
final_recommendations_df
```

```python
# Create explainable recommendations

explainable_recommendations = []

for student in final_similarity_df.index:

    # Get the top 3 faculty matches
    top_3 = final_similarity_df.loc[student].sort_values(
        ascending=False
    ).head(3)

    for rank, (faculty, final_score) in enumerate(top_3.items(), start=1):

        # Get individual component scores
        research_score = research_similarity_df.loc[student, faculty]

        academic_score = academic_similarity[
            students_df["name"].tolist().index(student),
            faculty_df["name"].tolist().index(faculty)
        ]

        goal_score = goal_similarity[
            students_df["name"].tolist().index(student),
            faculty_df["name"].tolist().index(faculty)
        ]

        # Get publication similarity score
        publication_score = publication_similarity_df.loc[
            student, faculty
        ]

        explanation = (
            f"Research similarity: {research_score:.3f}, "
            f"Academic similarity: {academic_score:.3f}, "
            f"Program goal similarity: {goal_score:.3f}, "
            f"Publication similarity: {publication_score:.3f}"
        )

        explainable_recommendations.append({
            "Student": student,
            "Rank": rank,
            "Recommended Faculty": faculty,
            "Research Score": round(research_score, 3),
            "Academic Score": round(academic_score, 3),
            "Goal Score": round(goal_score, 3),
            "Publication Match Score": round(publication_score, 3),
            "Final Match Score": round(final_score, 3),
            "Explanation": explanation
        })


# Convert to DataFrame
explainable_recommendations_df = pd.DataFrame(
    explainable_recommendations
)

# Display results
explainable_recommendations_df
```

# Limitations, Ethical Considerations, and Evaluation Plan

## Limitations

This project is a prototype student–faculty matching system designed to identify potential research-alignment opportunities between prospective graduate students and faculty members using text-based similarity.

The current implementation uses a small sample dataset containing five students and five faculty members. Therefore, the results are intended for demonstration purposes only and should not be interpreted as actual admission decisions, advisor assignments, or faculty-supervision decisions.

The quality of the recommendations depends on the quality, completeness, and consistency of the input data. The current student profiles include academic background, research interests, and graduate program goals. The faculty profiles include department, research interests, faculty focus areas, and publication information.

A larger real-world version could include additional information such as complete publication records, research-lab descriptions, student statements of purpose, relevant coursework, previous research experience, project portfolios, and other appropriately approved academic information.

The current weighted scoring model uses the following prototype weights:

- **Research interests: 55%**
- **Academic background: 15%**
- **Program goals: 15%**
- **Faculty publications: 15%**

These weights are manually selected prototype assumptions. Research interests receive the highest weight because direct research alignment is considered the primary factor in the current system. The weights have not yet been validated using historical matching data or feedback from faculty members and graduate program staff.

The system uses TF-IDF vectorization and cosine similarity to measure alignment between student and faculty profiles. These methods are effective for identifying overlapping words and phrases, but they may not fully capture deeper semantic relationships between related concepts.

The current prototype also does not consider operational factors that may affect whether a faculty match is feasible, such as faculty availability, advising workload, laboratory capacity, funding availability, or willingness to supervise new students.

## Ethical Considerations

This system is designed as a **decision-support tool** and should not be used as the sole basis for graduate admission decisions, faculty assignments, or research-supervision decisions. Human review should remain part of the final decision-making process.

Automated matching systems may introduce or reinforce bias if the underlying student or faculty data is incomplete, inaccurate, unrepresentative, or influenced by historical biases. Differences in terminology, writing style, academic background, or access to research opportunities may also affect similarity scores.

Sensitive personal information should not be unnecessarily included in the matching algorithm. When working with real student or applicant data, appropriate privacy, security, authorization, and university data-governance requirements should be followed.

The recommendations should remain transparent and explainable. For this reason, the system provides separate scores for:

- Research-interest alignment
- Academic-background alignment
- Program-goal alignment
- Publication alignment

These component scores help users understand why a faculty member was recommended instead of relying only on a single final score.

A low similarity score should not be interpreted as a measure of a student's academic ability, potential, or likelihood of success. It represents only the degree of alignment between the available text-based information in the student and faculty profiles.

## Evaluation Plan

A future version of this system should be evaluated using appropriately approved real-world data and expert feedback.

One possible evaluation approach would involve faculty members or graduate program staff reviewing student profiles and identifying faculty members they consider relevant matches. These expert-reviewed matches could then be compared with the system's ranked recommendations.

Possible evaluation methods include:

- **Precision@K:** Measures how many of the top K recommended faculty members are considered relevant.
